```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
```

```python
df=pd.read_excel("Matplotlib_seaborn_resources/linechart.xlsx")
```

```python
df
```

```python
plt.figure(figsize=(12,4))
plt.plot(df['Quarter'],df['Fridge'])
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(df['Quarter'],df['Fridge'])
plt.plot(df['Quarter'],df['Dishwasher'])
plt.plot(df['Quarter'],df['Washing Machine'])
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(df['Quarter'],df['Fridge'], label="Fridge")
plt.plot(df['Quarter'],df['Dishwasher'], label="Dishwasher")
plt.plot(df['Quarter'],df['Washing Machine'], label="Washing Machine")
plt.title("Productwise sales")
plt.ylabel("Sales in Millions ₹")
plt.xlabel("Quarter time->")
plt.legend()
plt.show()
```
