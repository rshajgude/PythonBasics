```python
def group_function(df, idx, col):
    if 70<=df[col].loc[idx]<=90:
        return "70-90"
    elif 50<=df[col].loc[idx]<=70:
        return "50-70"
    else:
        return "others"
```

```python
g=df.groupby(lambda idx : group_function(df, idx, 'temperature'))
g
```

```python
for key, data in g:
    print("\nkey:", key)
    print("data:", data)
```
