```python
f=open("Test.txt", "r")
```

```python
for line in f:
    print(line)
```

```python
f.close()
```

```python
with open("Test.txt","r") as f:
    for line in f:
        print(line)
```

```python
with open("Test.txt","r") as f:
    lines=f.readlines()
    print(lines)
```

```python
with open("TestWrite.txt","w") as f:
    f.write("We are writing line 1.")
    f.writelines("Lets write line 2. \nline 3")
```
